```python
# notebook to filter full blacklist filtered data for generating predictions all in one place
# ie include indels and snps in same predictions
# will be used for:
# 1. updating reviewer response
# 2. updating Figures 1B, 1C, and 1E
# 3. updating precision recall curves
```

```python
# import packages
import pandas as pd
```

```python
# open blacklist filtered predictions
all_ukbb_gtex = pd.read_csv('../raw_data/all.gtex.traits.mpra.blacklist.filtered.v1.txt', sep = '\t', low_memory=False)
```

```python
all_ukbb_gtex.keys()
```

```python
# iterate through and filter for only that chromosome
for chrom in [f'chr{i}' for i in range(1,23)]:
    # filter for that chromosome
    chrom_vars = all_ukbb_gtex[all_ukbb_gtex['chromosome'] == chrom].filter(['chromosome', 'hg38_id', 'allele1', 'allele2'])
    # drop duplicates
    vars2pred = chrom_vars.drop_duplicates()
    # break out position and make final df
    finalTSV = pd.DataFrame({
        0 : vars2pred['chromosome'],
        1 : [i.split(':')[1] for i in vars2pred['hg38_id']],
        2 : vars2pred['hg38_id'],
        3 : vars2pred['allele1'],
        4 : vars2pred['allele2']
    })
    # save to disk
    finalTSV.to_csv(f'../processed_data/chrom_vcfs/{chrom}_ukbb_gtex_all_blacklist_filtered_v1.tsv', sep = '\t', index = False, header = None)
```
