```python
# import packages
import pandas as pd
import os
import seaborn as sns
from scipy import stats
import numpy as np
from collections import Counter
```

```python
# open mpac prediction df (plus parm activity)
mpac_preds = pd.read_csv('/projects/tewhey-lab/buttsj/Variant_Effects/revision_experiments/promoter_sat_mut_comp/processed_data/mpac.250bp.promoter.preds.with.annotations.parm.act.tsv', sep = '\t')
```

```python
# open promoterAI predictions
promoterAI_preds = pd.read_csv('/projects/tewhey-lab/buttsj/Variant_Effects/revision_experiments/promoter_sat_mut_comp/processed_data/PrimateAI_and_PromoterAI_scores.hg38.250bp.filtered.vcf',sep = '\t', comment = '#', header = None, low_memory=False)
# break out predictions into its own column
promoterAI_preds.loc[:, 'promoterAI_score'] = [float(i.split(';')[-1]) if i != 'nan' else 'nan' for i in promoterAI_preds[7]]
```

```python
# make a dictionary for adding scores to full DF
promoterAI_score_dict = dict(zip(promoterAI_preds[2],
                                 promoterAI_preds['promoterAI_score']))
```

```python
# add to DF
mpac_preds.loc[:, 'promoterAI_score'] = [promoterAI_score_dict.get(i) if i in promoterAI_score_dict.keys() else 'nan' for i in mpac_preds['merge_id']]
```

```python
# count the number of missing predictions in promoterAI
mpac_preds['promoterAI_score'].tolist().count('nan')
```

```python
# save file to disk
mpac_preds.to_csv('/projects/tewhey-lab/buttsj/Variant_Effects/revision_experiments/promoter_sat_mut_comp/processed_data/mpac.250bp.promoter.preds.with.annotations.parm.act.promoterAI.tsv', sep = '\t', index = False)
```
